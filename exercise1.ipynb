```python
df = df.drop('Comunidades y Ciudades Autónomas', axis=1)
```

```python
df_nueva=df[df["General, vivienda nueva y de segunda mano"]=='Vivienda nueva']
df_nueva["Periodo"] = df_nueva["Periodo"].astype(float)
df_nueva["Total"] = df_nueva["Total"].str.replace(",", ".", regex=False).astype(float)
df_nueva.head(100)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline 
```

```python
plt.plot(df_nueva["Periodo"], df_nueva["Total"])
```
